```python
x = [random.randint(5, 10) for _ in range(10)]
y = [random.randint(10, 20) for _ in range(10)]
x2 = [i**2 for i in x]
y2 = [i**2 for i in y]
X = sm.add_constant(x)
res_ols = sm.OLS(y, X).fit()
print(f"sum:{sum(x)}, sumY:{sum(y)}, sum(X*Y):{sum(np.multiply(x,y))}, sumX**2:{sum(x2)}, sumY**2:{sum(y2)}, beta:{res_ols.params[1].round(2)} ")
```

```python
print(res_ols.summary())
```

```python
airfare = wooldridge.data('airfare')
```

```python
model_1 = smf.ols(formula = "fare~np.log(dist)", data = airfare[1:25]).fit()
model_2 = smf.ols(formula = "np.log(fare)~np.log(dist)", data = airfare[1:25]).fit()
model_3 = smf.ols(formula = "np.log(fare)~dist", data = airfare[1:25]).fit()
model_4 = smf.ols(formula = "fare~dist", data = airfare[1:25]).fit()
print(summary_col([model_1, model_2, model_3,model_4],
                  stars=True,
                  model_names=["Модель 1", "Модель 2", "Модель 3", "Модель 3"]))
```
